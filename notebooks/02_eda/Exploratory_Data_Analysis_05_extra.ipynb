#### Feature plot during capture duration

```python
# Load the uploaded dataset
file_path = '../Datasets/Per_UE_Datasets_final/ue_8609960480859058_normal_labeled.csv'
data = pd.read_csv(file_path)

output_dir = '../TimeSeriePlots/UE8609960480859058/'
os.makedirs(output_dir, exist_ok=True)  # Create directory if it doesn't exist

# Display the first few rows of the dataset to understand its structure
data.head(), data.info()


# Convert `_time` to datetime
data['_time'] = pd.to_datetime(data['_time'])

# Select key features for analysis
key_features = [
    'bearer_0_dl_total_bytes', 'bearer_0_ul_total_bytes', 'bearer_1_dl_total_bytes',
    'bearer_1_ul_total_bytes', 'dl_bitrate', 'ul_bitrate', 'ul_retx', 'ul_err',
    'ul_mcs', 'ul_path_loss', 'ul_phr', 'dl_err', 'dl_mcs', 'dl_retx', 'dl_tx',
    'cqi', 'epre', 'p_ue', 'pusch_snr', 'turbo_decoder_avg', 'ul_tx'
]



for feature in key_features:
    plt.figure(figsize=(12, 6))
    
    # Plot normal data points
    plt.plot(data['_time'], data[feature], label=f'{feature} (Normal)', alpha=0.7, linewidth=1.5, color='blue')
    
    # Highlight attack data points
    plt.scatter(data['_time'][data['binary_label'] == 1], 
                data[feature][data['binary_label'] == 1], 
                color='red', label='Attack', s=10, alpha=0.6)
    
    plt.title(f'{feature} Over Time with Attack Highlights', fontsize=14)
    plt.xlabel('Time', fontsize=12)
    plt.ylabel(feature, fontsize=12)
    plt.legend()
    plt.grid(True)
    #plt.show()
    
    # Save the plot to the directory
    file_path = os.path.join(output_dir, f'{feature}_over_time.png')
    plt.savefig(file_path)
    plt.close()
```
