```python
from tabulate import tabulate
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Read all CSVs from folder and concatenate
def read_data(folder_path):
    csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
    df_list = [pd.read_csv(os.path.join(folder_path, f)) for f in csv_files]
    return pd.concat(df_list, ignore_index=True)

# Define path and features
folder_path = '../Datasets/Per_Attack_Datasets'
features = ['epre','pusch_snr','p_ue','ul_mcs','cqi','ul_bitrate',
            'dl_mcs','dl_retx','ul_tx','dl_tx','ul_retx','dl_bitrate','dl_err','ul_err']

# Load and normalize
df = read_data(folder_path)
scaler = MinMaxScaler()
df_normalized = df.copy()
df_normalized[features] = scaler.fit_transform(df[features])

# Split by label
normal_df = df_normalized[df_normalized['binary_label'] == 0]
attack_df = df_normalized[df_normalized['binary_label'] == 1]

# Compute stats
table = []
for feature in features:
    normal_mean = round(normal_df[feature].mean(), 5)
    normal_std = round(normal_df[feature].std(), 5)
    attack_mean = round(attack_df[feature].mean(), 5)
    attack_std = round(attack_df[feature].std(), 5)
    table.append([feature, normal_mean, normal_std, attack_mean, attack_std])

# Tabulate the result
headers = ["Feature", "Normal Mean", "Normal Std", "Attack Mean", "Attack Std"]
table_str = tabulate(table, headers=headers, tablefmt="grid")

# Save to file
output_path = "normalized_feature_stats_tabulated.txt"
with open(output_path, "w") as f:
    f.write(table_str)

output_path
```
