```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
# Load the newly uploaded dataset
binary_stats_file = '../EDA_data/binary_stats_combined.csv'
binary_stats_data = pd.read_csv(binary_stats_file)

# Display the first few rows and structure of the dataset
binary_stats_data.head(), binary_stats_data.info()
```

```python
# Filter data for Label 0 and Label 1
label_0 = binary_stats_data[binary_stats_data['Unnamed: 0'] == 'Label 0']
label_1 = binary_stats_data[binary_stats_data['Unnamed: 0'] == 'Label 1']

# Extract features and means for both labels
features = label_0['Unnamed: 1']
label_0_means = label_0['mean']
label_1_means = label_1['mean']

# Plot grouped bar chart
x = range(len(features))
plt.figure(figsize=(12, 6))
plt.bar(x, label_0_means, width=0.4, label='Label 0', align='center')
plt.bar([i + 0.4 for i in x], label_1_means, width=0.4, label='Label 1', align='center')
plt.xticks([i + 0.2 for i in x], features, rotation=90)
plt.title('Mean Comparison Between Label 0 and Label 1')
plt.xlabel('Features')
plt.ylabel('Mean')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Filter data for Label 0 and Label 1
label_0 = binary_stats_data[binary_stats_data['Unnamed: 0'] == 'Label 0']
label_1 = binary_stats_data[binary_stats_data['Unnamed: 0'] == 'Label 1']

# Get the list of features
features = label_0['Unnamed: 1']

# Create boxplots for each feature
for feature in features:
    feature_label_0 = label_0[label_0['Unnamed: 1'] == feature]
    feature_label_1 = label_1[label_1['Unnamed: 1'] == feature]

    # Prepare box plot data
    box_data = [
        [feature_label_0['min'].values[0], feature_label_0['25%'].values[0],
         feature_label_0['50%'].values[0], feature_label_0['75%'].values[0],
         feature_label_0['max'].values[0]],
        [feature_label_1['min'].values[0], feature_label_1['25%'].values[0],
         feature_label_1['50%'].values[0], feature_label_1['75%'].values[0],
         feature_label_1['max'].values[0]],
    ]

    plt.figure(figsize=(8, 6))
    plt.boxplot(box_data, labels=['Label 0', 'Label 1'], showmeans=True)
    plt.title(f'Boxplot for {feature}')
    plt.ylabel('Value')
    plt.grid(True)
    plt.show()
```

```python
# Preparing the smaller table to show mean, std, and the percentage difference between Label 0 and Label 1

# Define the features, mean, and std values for both labels
# "bearer_0_dl_total_bytes", "bearer_0_ul_total_bytes", "bearer_1_dl_total_bytes",  "bearer_1_ul_total_bytes", 
features = [ "dl_bitrate", "ul_bitrate", "ul_retx", "ul_err", 
    "ul_mcs", "ul_path_loss", "ul_phr", "dl_err", "dl_mcs", "dl_retx", 
    "dl_tx", "cqi", "epre", "p_ue", "pusch_snr", "turbo_decoder_avg", "ul_tx"
]
# 6799738888.84, 29853010769.19, 6914896072.02, 40677478717.37,
# Values for Label 0 (Mean and Std)
label_0_mean = [
     1156557.91, 5323119.85,
    56.67, 0.0978, 18.43, 73.99, 24.69, 0.0557, 19.33, 35.01, 592.88, 12.15, -103.97, 
    -8.78, 24.77, 2.56, 608.60
]
# 17763902926.13, 80579429065.00, 18027516051.31, 112949432101.05,
label_0_std = [
    1573068.85, 
    8441077.29, 74.79, 0.6863, 7.55, 11.26, 12.26, 0.4083, 7.57, 57.39, 799.94, 3.26, 
    15.14, 12.07, 9.08, 0.56, 688.99
]

# Values for Label 1 (Mean and Std)
# 8383214.30, 1226116.81, 14839168.32, 6894213293.85,
label_1_mean = [
     45718.69, 6289109.28, 
    106.92, 2.94, 16.01, 76.05, 25.37, 0.36, 15.79, 41.62, 288.56, 10.28, 
    -110.94, -9.80, 21.93, 2.49, 780.64
]
# 3494324.02, 498769.36, 7014418.22, 6686942321.53,
label_1_std = [
    35176.91, 6826681.10, 
    86.46, 10.90, 7.53, 11.82, 11.40, 1.26, 9.37, 48.99, 158.50, 3.97, 
    11.62, 8.02, 6.67, 0.49, 383.15
]

# Calculate percentage difference for means
percentage_diff_mean = [(l1 - l0) / abs(l0) * 100 for l0, l1 in zip(label_0_mean, label_1_mean)]

# Create the summarized table
summary_table = pd.DataFrame({
    "Feature": features,
    "Label 0 Mean": label_0_mean,
    "Label 0 Std": label_0_std,
    "Label 1 Mean": label_1_mean,
    "Label 1 Std": label_1_std,
    "Mean Difference (%)": percentage_diff_mean
})
```

```python
summary_table
```

```python
# Formatting the numbers in the table for better readability
summary_table_formatted = summary_table.copy()

# Formatting mean and std values to scientific notation with 2 decimals
summary_table_formatted["Label 0 Mean"] = summary_table["Label 0 Mean"].map("{:.2e}".format)
summary_table_formatted["Label 0 Std"] = summary_table["Label 0 Std"].map("{:.2e}".format)
summary_table_formatted["Label 1 Mean"] = summary_table["Label 1 Mean"].map("{:.2e}".format)
summary_table_formatted["Label 1 Std"] = summary_table["Label 1 Std"].map("{:.2e}".format)

# Formatting mean difference percentage to 2 decimal places
summary_table_formatted["Mean Difference (%)"] = summary_table["Mean Difference (%)"].map("{:.2f}".format)

# Display the formatted table
display(summary_table_formatted)
```
